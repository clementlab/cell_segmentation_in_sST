# Identifying good regions to focus on

```python
import os
from tifffile import imread, imwrite
from matplotlib import pyplot as plt
import matplotlib.patches as patches
from matplotlib_scalebar.scalebar import ScaleBar


os.makedirs('supplementals', exist_ok=True)
```

# Kidney
12000-14000

```python
kid_img = imread('/uufs/chpc.utah.edu/common/home/u1531817/20250523_cell_seg_eval/data/mouse_kidney/Visium_HD_Mouse_Kidney_tissue_image.btf')
microns_per_pixel = 0.27390206550511054
plt.figure(figsize=(10, 10))
plt.subplot(1, 2, 1)
plt.imshow(kid_img, origin='lower')
plt.title('Full Kidney Image')

rect = patches.Rectangle(
    (12000, 12000), 
    2000,
    2000,
    linewidth=2,
    edgecolor='red',
    facecolor='none'
)

plt.gca().add_patch(rect)

scalebar = ScaleBar(
    dx=microns_per_pixel,
    units='µm',
    location='lower right',
    scale_loc='bottom',
    color='black',
    box_alpha=0.7) 
plt.gca().add_artist(scalebar)

plt.subplot(1, 2, 2)
plt.imshow(kid_img, origin='lower')
plt.xlim(12000, 14000)
plt.ylim(12000, 14000)
scalebar = ScaleBar(
    dx=microns_per_pixel,
    units='µm',
    location='lower right',
    scale_loc='bottom',
    color='black',
    box_alpha=0.7) 
plt.gca().add_artist(scalebar)

plt.title('Kidney Region of Interest')
plt.savefig('supplementals/kidney_image.png', dpi=300)
```
